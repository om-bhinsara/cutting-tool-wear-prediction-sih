# Image Preprocessing & Wear Progression

Visual evidence notebook for the tool-wear benchmark.

```python
from pathlib import Path
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

ROOT=Path("..").resolve()
df=pd.read_csv(ROOT/"labels.csv")
s=1
rows=df[df.Set==s].iloc[[0,10,20,30,40,50,59]]

fig,axes=plt.subplots(1,len(rows),figsize=(18,3))
for ax,(_,r) in zip(axes,rows.iterrows()):
    ax.imshow(Image.open(ROOT/"edge_images"/r.ImageName))
    ax.set_title(f'{r.wear_um:.0f} µm\n{r.stage_name}')
    ax.axis("off")
plt.tight_layout()
plt.show()
```
